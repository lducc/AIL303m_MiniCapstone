```python
import pandas as pd

train = pd.read_csv("/kaggle/input/dataset2/test_pca.csv")
test = pd.read_csv("/kaggle/input/dataset2/train_pca.csv") 

train.head()
test.head()
```

```python
from sklearn.model_selection import train_test_split

x_test = test.drop([
    'Heart_Disease_Risk', 'Age', 'Height_cm', 'Weight_kg', 
    'BMI', 'Systolic_BP', 'Diastolic_BP', 
    'Cholesterol_Total', 'Cholesterol_HDL',
], axis=1)

x_train = train.drop([
    'Heart_Disease_Risk', 'Age', 'Height_cm', 'Weight_kg', 
    'BMI', 'Systolic_BP', 'Diastolic_BP', 
    'Cholesterol_Total', 'Cholesterol_HDL',
], axis=1)

y_train = train['Heart_Disease_Risk']
y_test = test['Heart_Disease_Risk']
```

```python
from sklearn import svm
from sklearn.metrics import accuracy_score

model = svm.SVC()
model.fit(x_train, y_train)
predictions = model.predict(x_test)
accuracy = accuracy_score(y_test, predictions)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
import pickle

SVM_filename = "heart-care-svm.pkl"  

with open(SVM_filename, 'wb') as file:  
    pickle.dump(model, file)
```
